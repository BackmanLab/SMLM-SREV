```python
filepath = ''
EdU_RD = scipy.io.loadmat(filepath + 'Domain04.mat')
filepath = ''
BrdU_RD = scipy.io.loadmat(filepath + 'Domain04.mat')
filepath = ''
Nucleosome_RD = scipy.io.loadmat(filepath + 'Domain04.mat')
```

```python
plt.figure(figsize=(10, 6))
plt.plot(EdU_RD['meanRD'][0]/EdU_RD['meanRD'][0].max(), label='EdU', color='forestgreen', linewidth=5)
plt.plot(BrdU_RD['meanRD'][0]/BrdU_RD['meanRD'][0].max(), label='BrdU', color='firebrick', linewidth=5)
plt.plot(Nucleosome_RD['meanRD'][0]/Nucleosome_RD['meanRD'][0].max(), label='Nucleosome', color='goldenrod', linewidth=5)

# Enhance the plot with titles, labels, legend, and grid
plt.title("Normalized Radial Density", fontsize=26)
plt.xlabel("Radius [nm]", fontsize=24)
plt.tick_params(axis='x', labelsize=20)
plt.tick_params(axis='y', labelsize=20)
# plt.ylabel("", fontsize=14)
plt.legend(fontsize=20)
plt.savefig("", bbox_inches="tight")

plt.plot()
```
